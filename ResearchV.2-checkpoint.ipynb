# 3. Cleaning

```python
df = df.drop_duplicates()

df['Timestamp'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])

df = df.sort_values(by='Timestamp').reset_index(drop=True)

df = df.drop(columns=['Date', 'Time'])

print(f"Dataframe shape after initial cleanup: {df.shape}")
```

```python
df = df[df['Temperature'] <= 55]

df = df[df['PM2.5'] <= 1000]
```

```python
df['Date'] = pd.to_datetime(df['Timestamp'])
df = df.sort_values(['Station', 'Date']).reset_index(drop=True)

# Precipitation: only use 0 if missing truly means no rain recorded
df['Precipitation_missing'] = df['Precipitation'].isna().astype(int)
df['Precipitation'] = df['Precipitation'].fillna(0)

features_to_impute = ['Humidity', 'Temperature', 'PM2.5', 'AQI']

# Missing indicators
for col in features_to_impute:
    df[col + '_missing'] = df[col].isna().astype(int)

# Time-based imputation within each station
for col in features_to_impute:
    df[col] = (
        df.groupby('Station', group_keys=False)
          .apply(lambda g: g.set_index('Date')[col].interpolate(method='time', limit=3))
          .reset_index(level=0, drop=True)
    )
    df[col] = df.groupby('Station')[col].ffill(limit=2)
    df[col] = df.groupby('Station')[col].bfill(limit=2)

# Final fallback: station median for any remaining gaps
for col in features_to_impute:
    df[col] = df.groupby('Station')[col].transform(lambda x: x.fillna(x.median()))

# Optional final fallback if any station median is still NaN
for col in features_to_impute:
    df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

#### Merged Date/Time to Timestamp, removed duplicates and implausible outliers, and sorted by station/time. Recorded precipitation missingness then filled NaNs with 0. For Humidity, Temperature, PM2.5, and AQI added missing indicators, applied per‑station time interpolation (limit=3) with short ffill/bfill (limit=2), then filled remaining gaps with station medians and a global median fallback. Printed remaining missing counts.

# 4 Feature Engineering

```python
import pandas as pd
import numpy as np

# =====================================================
# 1. SORT
# =====================================================
df = df.copy()

df["Timestamp"] = pd.to_datetime(df["Timestamp"])

df = (
    df.sort_values(["Station", "Timestamp"])
      .reset_index(drop=True)
)

# =====================================================
# 2. DATETIME FEATURES
# =====================================================
df["hour"] = df["Timestamp"].dt.hour
df["dayofweek"] = df["Timestamp"].dt.dayofweek
df["month"] = df["Timestamp"].dt.month
df["day"] = df["Timestamp"].dt.day
df["is_weekend"] = (df["dayofweek"] >= 5).astype(int)

# =====================================================
# 3. TARGET (1 hour ahead)
# =====================================================
FORECAST_HORIZON = 1

df["Target_PM2.5"] = (
    df.groupby("Station")["PM2.5"]
      .shift(-FORECAST_HORIZON)
)

# =====================================================
# 4. LAG FEATURES
# =====================================================
lag_features = [
    "PM2.5",
    "AQI",
    "Temperature",
    "Humidity",
    "Precipitation"
]

for feature in lag_features:
    for lag in [1, 3, 24]:
        df[f"{feature}_lag_{lag}"] = (
            df.groupby("Station")[feature]
              .shift(lag)
        )

# =====================================================
# 5. ROLLING FEATURES
# =====================================================

def add_rolling_features(column):

    g = df.groupby("Station")[column]

    df[f"{column}_roll_mean_3"] = g.transform(
        lambda x: x.shift(1).rolling(3, min_periods=1).mean()
    )

    df[f"{column}_roll_mean_24"] = g.transform(
        lambda x: x.shift(1).rolling(24, min_periods=1).mean()
    )

    df[f"{column}_roll_std_24"] = g.transform(
        lambda x: x.shift(1).rolling(24, min_periods=2).std()
    )

# PM2.5
add_rolling_features("PM2.5")

# AQI
add_rolling_features("AQI")

# Temperature
add_rolling_features("Temperature")

# Humidity
add_rolling_features("Humidity")

# Extra PM2.5 features
g_pm = df.groupby("Station")["PM2.5"]

df["PM2.5_roll_std_6"] = g_pm.transform(
    lambda x: x.shift(1).rolling(6, min_periods=2).std()
)

df["PM2.5_roll_max_24"] = g_pm.transform(
    lambda x: x.shift(1).rolling(24, min_periods=1).max()
)

# =====================================================
# 6. INTERACTION FEATURES
# =====================================================
df["PM2.5_lag1_x_humidity"] = (
    df["PM2.5_lag_1"] * df["Humidity"]
)

df["PM2.5_lag1_x_temperature"] = (
    df["PM2.5_lag_1"] * df["Temperature"]
)

df["PM2.5_lag1_div_precip"] = (
    df["PM2.5_lag_1"] /
    (df["Precipitation"] + 1)
)

df["Temp_Humidity"] = (
    df["Temperature"] *
    df["Humidity"]
)

# =====================================================
# 7. SEASON
# =====================================================
season_map = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Autumn", 10: "Autumn", 11: "Autumn"
}

df["Season"] = df["month"].map(season_map)

# =====================================================
# 8. TRAFFIC FEATURES
# =====================================================
df["Rush_Hour"] = (
    ((df["hour"] >= 7) & (df["hour"] <= 9)) |
    ((df["hour"] >= 17) & (df["hour"] <= 20))
).astype(int)

df["Night"] = (
    (df["hour"] >= 0) &
    (df["hour"] < 6)
).astype(int)

df["Weekend_Rush"] = (
    (df["Rush_Hour"] == 1) &
    (df["is_weekend"] == 1)
).astype(int)

# =====================================================
# 9. CYCLICAL FEATURES
# =====================================================
df["hour_sin"] = np.sin(2*np.pi*df["hour"]/24)
df["hour_cos"] = np.cos(2*np.pi*df["hour"]/24)

df["month_sin"] = np.sin(2*np.pi*df["month"]/12)
df["month_cos"] = np.cos(2*np.pi*df["month"]/12)

# =====================================================
# 10. REMOVE NaNs CREATED BY SHIFTING
# =====================================================
df = df.dropna().reset_index(drop=True)

# =====================================================
# 11. ONE-HOT ENCODE
# =====================================================
df = pd.get_dummies(
    df,
    columns=["Station", "Season"],
    drop_first=False
)

# =====================================================
# 12. DROP UNUSED COLUMNS
# =====================================================
df.drop(
    columns=["Timestamp", "Date", "Time"],
    inplace=True,
    errors="ignore"
)

print("=" * 60)
print("Feature Engineering Completed Successfully!")
print("Final Shape:", df.shape)
print("=" * 60)

display(df.head())
```

#### Extracted time features, created a 1-hour-ahead target, and engineered temporal, lag, rolling, interaction, seasonal, traffic, and cyclical features per station in a time-aware way. Specifically: sorted by station/timestamp; derived hour/day/month/weekend flags; shifted PM2.5 to form the forecast target; added lags (1, 3, 24) for key signals; computed rolling means/std/max with past-only windows to capture short- and daily-history; built interactions between recent PM2.5, humidity, temperature, and precipitation; mapped month → season; added rush-hour/night/weekend-rush indicators; encoded cyclical hour/month with sine/cosine; dropped rows with NaNs from shifting; one-hot encoded Station and Season; and removed unused columns. Result: a modeling-ready dataset with temporal context, short- and long-term history, seasonality, and interaction predictors.

```python
import time
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.model_selection import cross_val_score


def evaluate_model(model, X_train, X_test, y_train, y_test, model_name):

    start = time.time()

    model.fit(X_train, y_train)

    filename = model_name.replace(" ", "_") + ".pkl"
    joblib.dump(model, filename)

    fit_time = time.time() - start

    # Predictions
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    # Metrics
    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))

    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)

    test_mae = mean_absolute_error(y_test, test_pred)

    # Cross Validation RMSE
    cv_rmse = np.sqrt(
        -cross_val_score(
            model,
            X_train,
            y_train,
            scoring="neg_mean_squared_error",
            cv=5
        ).mean()
    )

    print(f"\n=== {model_name} ===")
    print(f"Train RMSE: {train_rmse:.3f} | Test RMSE: {test_rmse:.3f} | CV RMSE: {cv_rmse:.3f}")
    print(f"Test MAE: {test_mae:.3f} | Train R²: {train_r2:.4f} | Test R²: {test_r2:.4f}")
    print(f"Fit Time: {fit_time:.3f} sec")

    # Plot
    plt.figure(figsize=(6,6))

    plt.scatter(y_test, test_pred, alpha=0.6)

    mn = min(y_test.min(), test_pred.min())
    mx = max(y_test.max(), test_pred.max())

    plt.plot(
        [mn, mx],
        [mn, mx],
        'r--',
        linewidth=2,
        label="Perfect prediction"
    )

    plt.xlabel("Actual PM2.5")
    plt.ylabel("Predicted PM2.5")

    plt.title(f"{model_name} - Actual vs Predicted (R²={test_r2:.3f})")

    plt.legend()

    plt.grid(alpha=0.3)

    plt.show()

    return {
        "Model": model_name,
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "CV RMSE": cv_rmse,
        "Train R²": train_r2,
        "Test R²": test_r2,
        "Test MAE": test_mae,
        "Fit Time": fit_time
    }
```

```python
import joblib
import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

def load_and_evaluate(model_name, scaled=True):

    print("="*60)
    print(f"Loading {model_name}...")

    filename = model_name.replace(" ", "_") + ".pkl"

    model = joblib.load(filename)

    if scaled:
        Xtr = X_train_scaled
        Xte = X_test_scaled
    else:
        Xtr = X_train
        Xte = X_test

    train_preds = model.predict(Xtr)
    test_preds = model.predict(Xte)

    metrics = {
        "Model": model_name,
        "Train RMSE": np.sqrt(mean_squared_error(y_train, train_preds)),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, test_preds)),
        "Train R²": r2_score(y_train, train_preds),
        "Test R²": r2_score(y_test, test_preds),
        "Test MAE": mean_absolute_error(y_test, test_preds),
    }

    results.append(metrics)
    preds_store[model_name] = test_preds

    print(f"✓ {model_name} loaded.")

    return model
```

# 7. Models - Tuned

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import pandas as pd
import time
import joblib

tuned_results = []
tuned_preds = {}
tuned_models = {}

def tune_model(model_name, model, param_grid, scaled=True):

    print("="*70)
    print(f"Tuning {model_name}...")

    if scaled:
        Xtr = X_train_scaled
        Xte = X_test_scaled
    else:
        Xtr = X_train
        Xte = X_test

    t0 = time.time()

    grid = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        cv=3,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1,
        verbose=0
    )

    grid.fit(Xtr, y_train)

    fit_time = time.time() - t0

    best_model = grid.best_estimator_

    tuned_models[model_name] = best_model

    filename = model_name.replace(" ", "_") + "_Tuned.pkl"
    joblib.dump(best_model, filename)

    train_preds = best_model.predict(Xtr)
    test_preds = best_model.predict(Xte)

    tuned_preds[model_name] = test_preds

    cv_rmse = np.sqrt(
        -grid.best_score_
    )

    metrics = {
        "Model": model_name,
        "Train RMSE": np.sqrt(mean_squared_error(y_train, train_preds)),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, test_preds)),
        "CV RMSE": cv_rmse,
        "Train R²": r2_score(y_train, train_preds),
        "Test R²": r2_score(y_test, test_preds),
        "Test MAE": mean_absolute_error(y_test, test_preds),
        "Fit Time (sec)": fit_time
    }

    tuned_results.append(metrics)

    print("\nBest Parameters:")
    print(grid.best_params_)

    display(pd.DataFrame([metrics]).round(4))

    print(f"✓ {model_name} tuned.\n")

    pd.DataFrame(tuned_results).to_csv(
    "tuned_results.csv",
    index=False
)

    return best_model
```

```python
import joblib
import numpy as np
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

tuned_results = []
tuned_preds = {}

def load_tuned_model(model_name, scaled=True):

    print("="*60)
    print(f"Loading {model_name}...")

    if scaled:
        Xtr = X_train_scaled
        Xte = X_test_scaled
    else:
        Xtr = X_train
        Xte = X_test

    filename = model_name.replace(" ", "_") + "_Tuned.pkl"

    model = joblib.load(filename)

    train_preds = model.predict(Xtr)
    test_preds = model.predict(Xte)

    metrics = {
        "Model": model_name,
        "Train RMSE": np.sqrt(mean_squared_error(y_train, train_preds)),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, test_preds)),
        "Train R²": r2_score(y_train, train_preds),
        "Test R²": r2_score(y_test, test_preds),
        "Test MAE": mean_absolute_error(y_test, test_preds)
    }

    tuned_results.append(metrics)
    tuned_preds[model_name] = test_preds

    print(f"✓ Loaded {filename}")

    return model
```
